```python
import os
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
os.environ["OPENAI_ROUTER_API_KEY"]=os.getenv("OPENAI_ROUTER_API_KEY")
```

```python
from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model="google/gemini-2.5-flash", # Router-supported model identifier
    api_key=os.getenv("OPENAI_ROUTER_API_KEY"),       # Your router/OpenRouter key
    base_url="https://openrouter.ai/api/v1",
      model_kwargs={
        "extra_body": {
            "models": [
                "qwen/qwen3.8-27b:free",
                "meta-llama/llama-3.3-70b-instruct:free",
                "google/gemini-2.5-flash:free"
            ]
        }
    },  # Or your custom proxy URL
)
response=llm.invoke("How can you help me")
response
```
