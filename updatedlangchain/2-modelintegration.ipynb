```python
import os
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("gpt-4.1")
model
```

```python
import json
response=model.invoke("Hello How are You ")
print(json.dumps(response.content, indent=2))
```

```python
from langchain.chat_models import init_chat_model
model=init_chat_model("google_genai:gemini-3.5-flash-lite")
response=model.invoke("Why do parrots talk")
print(json.dumps(response.content, indent=2))
```

```python
from langchain_openai import ChatOpenAI
model=ChatOpenAI(model="gpt-4.1")
response=model.invoke("Hello How are You")
print(json.dumps(response.content, indent=2))
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model=ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response=model.invoke("Why do parrots talk")
print(json.dumps(response.content, indent=2))
```

```python
#Groq Model Integration 
model=init_chat_model("groq:qwen/qwen3.6-27b")
response=model.invoke("Why do parrots talk")
print(json.dumps(response.content, indent=2))
```

```python
from langchain_groq import ChatGroq
model = ChatGroq(
    model="qwen/qwen3.8-27b",
     reasoning_effort="none",
     temperature=0.7,
    top_p=0.80,
    reasoning_format="hidden"
)
response=model.invoke("Why do parrots talk")
print(json.dumps(response.content, indent=2))
```

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Write me a 200 words paragraph on Artificial Intelligence"):
    print(chunk.text,end="|",flush=True)
```

```python
for chunk in model.stream("Why do parrots have colorful feathers"):
    print(chunk.text,end="|",flush=True)
```

```python
model.invoke("Why do parrots have colorful feathers")
    
```

```python
responses=model.batch(["Why do parrots have colorful feathers","How do airplanes fly","what is quantum computing"],config={"max_concurrency": 1})
for response in responses:
    print(response.content)
    print("-" * 40)
```
