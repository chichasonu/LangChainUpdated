```python
import os
from langchain.chat_models import init_chat_model   
from pydantic import BaseModel, Field
from dotenv import load_dotenv
load_dotenv()
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
os.environ["GOOGLE_API_KEY"]=os.getenv("GOOGLE_API_KEY")
os.environ["GROQ_API_KEY"]=os.getenv("GROQ_API_KEY")
os.environ["OPENAI_ROUTER_API_KEY"]=os.getenv("OPENAI_ROUTER_API_KEY")
model=init_chat_model("gpt-4o-mini")
```

```python
class Movie(BaseModel):
    title: str=Field(description="The title of the movie")
    director: str=Field(description="The director of the movie")
    year: int=Field(description="The year the movie was released")
    genre: str=Field(description="The genre of the movie")
    rating: float=Field(description="The rating of the movie")
    description: str=Field(description="The description of the movie")
```

```python
from typing import List
 
class MovieList(BaseModel):
    movies: List[Movie] = Field(description="A list of movies")
```

```python
model_with_structure=model.with_structured_output(MovieList)
model_with_structure
```

```python
response=model_with_structure.invoke("Best Malayalam Movies in 2026 year")
print(response)
```

```python
import json
print(json.dumps(response.model_dump(), indent=2))
```

```python
from pydantic import BaseModel

class Actor(BaseModel):
    name: str
    role: str

class Movie(BaseModel):
    title: str=Field(description="The title of the movie")
    director: str=Field(description="The director of the movie")
    year: int=Field(description="The year the movie was released")
    genre: str=Field(description="The genre of the movie")
    rating: float=Field(description="The rating of the movie")
    description: str=Field(description="The description of the movie")
    budget: float=Field(description="The budget of the movie")
    cast: List[Actor]
```

```python
model_with_structure=model.with_structured_output(Movie)
model_with_structure
```

```python
response=model_with_structure.invoke("Tell about movie Aasha released in 2026")
import json
print(json.dumps(response.model_dump(), indent=2))
```
